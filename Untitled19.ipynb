```python
# ── CELDA 5: Tratamiento quirúrgico de outliers ───────────────

# Recargar dataset limpio (volvemos al punto tras imputar nulos)
df = pd.read_csv(ruta, sep='\t')
df['Income'] = df['Income'].fillna(df['Income'].median())

# 1. Year_Birth: eliminar edades imposibles (nacidos antes de 1924)
df = df[df['Year_Birth'] >= 1924]

# 2. Income: eliminar valor extremo
df = df[df['Income'] <= 200000]

filas_antes = 2240
print(f"Filas después de limpieza quirúrgica: {len(df)}")
print(f"Filas eliminadas: {filas_antes - len(df)}")
```

```python
# ── CELDA 6: Corrección de tipos de datos ────────────────────
# Dt_Customer está como object, debe ser datetime
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], dayfirst=True)

print(f"Tipo anterior: object")
print(f"Tipo actual: {df['Dt_Customer'].dtype}")
print(f"\nEjemplo de valores:")
print(df['Dt_Customer'].head())
```

```python
# ── CELDA 7: Ingeniería de características ────────────────────

# 1. Edad del cliente
df['Age'] = 2026 - df['Year_Birth']

# 2. Antigüedad como cliente en días
fecha_actual = pd.Timestamp('2026-04-10')
df['Seniority'] = (fecha_actual - df['Dt_Customer']).dt.days

# 3. Gasto total
df['Total_Spending'] = (df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] +
                        df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds'])

# 4. Total de compras por canal
df['Total_Purchases'] = (df['NumDealsPurchases'] + df['NumWebPurchases'] +
                         df['NumCatalogPurchases'] + df['NumStorePurchases'])

# 5. Total de hijos en casa
df['Total_Children'] = df['Kidhome'] + df['Teenhome']

# 6. Total campañas aceptadas
df['Total_Campaigns'] = (df['AcceptedCmp1'] + df['AcceptedCmp2'] + df['AcceptedCmp3'] +
                         df['AcceptedCmp4'] + df['AcceptedCmp5'])

# Verificar las nuevas columnas
print(f"Nuevas columnas creadas. Shape actual: {df.shape}")
print()
display(df[['Age', 'Seniority', 'Total_Spending',
            'Total_Purchases', 'Total_Children',
            'Total_Campaigns']].head())
```

```python
# ── CELDA 8: Eliminación de columnas redundantes o sin valor ──

# Year_Birth y Dt_Customer: reemplazadas por Age y Seniority
# Kidhome, Teenhome: resumidas en Total_Children
# AcceptedCmp1-5: resumidas en Total_Campaigns
# Mnt*: resumidas en Total_Spending
# Num*Purchases: resumidas en Total_Purchases
# ID: solo identificador, no aporta al análisis
# Z_CostContact, Z_Revenue: valor constante, sin variabilidad

columnas_eliminar = [
    'Year_Birth', 'Dt_Customer',
    'Kidhome', 'Teenhome',
    'AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5',
    'MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts',
    'MntSweetProducts', 'MntGoldProds',
    'NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases',
    'ID', 'Z_CostContact', 'Z_Revenue'
]

df = df.drop(columns=columnas_eliminar)

print(f"Shape final: {df.shape}")
print(f"\nColumnas restantes:")
print(df.columns.tolist())
```

```python
# ── CELDA 9: Explorar valores únicos de variables categóricas ─
# Antes de codificar, revisamos qué valores existen en cada variable
# para elegir la técnica correcta de codificación

print("=== EDUCATION ===")
print(df['Education'].value_counts())

print("\n=== MARITAL_STATUS ===")
print(df['Marital_Status'].value_counts())
```

```python
# ── CELDA 10: Limpieza y codificación de variables categóricas ─
# Marital_Status tiene valores inválidos (Alone, Absurd, YOLO)
# que no representan categorías reales — se eliminan (7 registros)
# Education tiene orden jerárquico — se aplica Label Encoding
# Marital_Status no tiene orden lógico — se aplica One-Hot Encoding

# 1. Eliminar registros con categorías inválidas
df = df[~df['Marital_Status'].isin(['Alone', 'Absurd', 'YOLO'])]
print(f"Registros tras limpiar Marital_Status: {len(df)}")

# 2. Label Encoding para Education (tiene orden jerárquico)
orden_education = {'Basic': 0, '2n Cycle': 1, 'Graduation': 2, 'Master': 3, 'PhD': 4}
df['Education'] = df['Education'].map(orden_education)

# 3. One-Hot Encoding para Marital_Status (sin orden lógico)
df = pd.get_dummies(df, columns=['Marital_Status'], prefix='MS', dtype=int)

print(f"\nShape final: {df.shape}")
print(f"\nColumnas resultantes:")
print(df.columns.tolist())
```

```python
# ── CELDA 11: Estadísticas descriptivas finales ───────────────
# Dataset ya limpio, transformado y con variables derivadas
# Ahora sí es válido calcular estadísticas representativas

print("=== VARIABLES NUMÉRICAS ===")
display(df.describe().round(2))
```

```python
# ── CELDA 12A: Distribución de variables numéricas (versión visual) ──
# Histogramas con KDE para visualizar distribución y forma de cada variable.
# Paleta de colores variada para facilitar la lectura en el reporte PDF.

import matplotlib.pyplot as plt
import seaborn as sns

variables = ['Income', 'Age', 'Total_Spending', 'Total_Purchases',
             'Seniority', 'Total_Children', 'Total_Campaigns', 'Recency']

colores = ['#3498db', '#e74c3c', '#2ecc71', '#f39c12',
           '#9b59b6', '#1abc9c', '#e67e22', '#e91e63']

fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(22, 10))
axes = axes.flatten()

for i, (col, color) in enumerate(zip(variables, colores)):
    sns.histplot(df[col], bins=30, kde=True, ax=axes[i],
                 color=color, edgecolor='white', alpha=0.75)
    axes[i].set_title(col, fontsize=12, fontweight='bold', pad=10)
    axes[i].set_xlabel('Valor', fontsize=9)
    axes[i].set_ylabel('Frecuencia', fontsize=9)
    axes[i].spines['top'].set_visible(False)
    axes[i].spines['right'].set_visible(False)

fig.patch.set_facecolor('#f8f9fa')
plt.suptitle('Distribución de Variables Numéricas',
             fontsize=16, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

```python
# ── CELDA 12B: Mapa de calor de correlaciones ─────────────────
# Muestra qué tan relacionadas están las variables numéricas entre sí.
# Correlaciones cercanas a 1 o -1 indican relación fuerte.
# Útil para identificar variables redundantes o patrones relevantes
# antes del clustering.

import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(14, 10))

correlaciones = df.corr(numeric_only=True)

mascara = sns.heatmap(
    correlaciones,
    annot=True,
    fmt='.2f',
    cmap='RdYlGn',
    center=0,
    linewidths=0.5,
    linecolor='white',
    annot_kws={'size': 8},
    ax=ax
)

ax.set_title('Mapa de Correlaciones', fontsize=15, fontweight='bold', pad=15)
fig.patch.set_facecolor('#f8f9fa')
plt.tight_layout()
plt.show()
```

```python
# ── CELDA 12C: Distribución de variables categóricas ──────────
# Gráficos de barras para Education y Marital_Status originales.
# Permiten visualizar la composición del dataset por perfil
# educativo y estado civil antes de la codificación.

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.patch.set_facecolor('#f8f9fa')

# Education
edu_labels = ['Basic', '2n Cycle', 'Graduation', 'Master', 'PhD']
edu_counts = df['Education'].value_counts().sort_index()
edu_counts.index = edu_labels
axes[0].bar(edu_labels, edu_counts.values,
            color=['#3498db','#2ecc71','#e74c3c','#f39c12','#9b59b6'],
            edgecolor='white', width=0.6)
axes[0].set_title('Distribución por Nivel Educativo', fontweight='bold', fontsize=12)
axes[0].set_ylabel('Cantidad de clientes')
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# Marital Status
ms_cols = ['MS_Married', 'MS_Together', 'MS_Single', 'MS_Divorced', 'MS_Widow']
ms_labels = ['Married', 'Together', 'Single', 'Divorced', 'Widow']
ms_counts = df[ms_cols].sum()
axes[1].bar(ms_labels, ms_counts.values,
            color=['#e74c3c','#3498db','#2ecc71','#f39c12','#9b59b6'],
            edgecolor='white', width=0.6)
axes[1].set_title('Distribución por Estado Civil', fontweight='bold', fontsize=12)
axes[1].set_ylabel('Cantidad de clientes')
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.suptitle('Distribución de Variables Categóricas', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ── CELDA 13: Normalización con MinMaxScaler ──────────────────
# Se normalizan variables numéricas con escalas distintas para
# prepararlas para clustering (K-Means usa distancias euclidianas,
# por lo que variables en escalas grandes dominarían el cálculo).
# Las variables binarias (0/1) NO se normalizan — ya están en la
# misma escala.

from sklearn.preprocessing import MinMaxScaler

# Columnas a normalizar (excluir binarias)
binarias = ['Complain', 'Response', 'MS_Divorced', 'MS_Married',
            'MS_Single', 'MS_Together', 'MS_Widow']

columnas_normalizar = [col for col in df.columns if col not in binarias]

scaler = MinMaxScaler()
df[columnas_normalizar] = scaler.fit_transform(df[columnas_normalizar])

print("Normalización completada.")
print(f"\nRango tras normalizar (debe ser 0 a 1):")
display(df[columnas_normalizar].describe().loc[['min','max']].round(4))
```
